```python
import librosa
import soundfile as sf
import numpy as np


def preprocess_audio(
    input_path,
    output_path,
    target_sr=22050
):
    """
    Load, convert to mono, resample,
    and save audio as PCM-16 WAV.
    """

    # Load at original sample rate
    waveform, original_sr = librosa.load(
        input_path,
        sr=None,
        mono=False
    )

    # -------------------------
    # Mono conversion
    # -------------------------
    if waveform.ndim == 2:
        waveform = np.mean(
            waveform,
            axis=0
        )

    # -------------------------
    # Validate
    # -------------------------
    if waveform.size == 0:
        raise ValueError(
            f"Empty audio: {input_path}"
        )

    # -------------------------
    # Resampling
    # -------------------------
    if original_sr != target_sr:

        waveform = librosa.resample(
            waveform,
            orig_sr=original_sr,
            target_sr=target_sr
        )

    # -------------------------
    # Save
    # -------------------------
    sf.write(
        output_path,
        waveform,
        target_sr,
        subtype="PCM_16"
    )

    return {
        "original_sr": original_sr,
        "target_sr": target_sr,
        "duration": len(waveform) / target_sr
    }
```
